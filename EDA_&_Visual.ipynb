```python
import pandas as pd
import matplotlib.pyplot as plt


pd.read_excel("/content/cleaned_data.xlsx")
df=pd.read_excel("/content/cleaned_data.xlsx")
df.drop(labels="Unnamed: 0", axis=1,inplace=True)

df["Package"]. value_counts() #To check package with most registration.
df.AgeRange.value_counts() #age range with most registration
df.LGA.value_counts() #LGA with most registration
df.groupby(["LGA","AgeRange"])["Package"].count() #total registration by age range from each LGA
df.groupby(["LGA","Package"])["AgeRange"].count() #each package registration breakdown by age in each LGA

df["submit_month"]=df["submit_time"].dt.month #extracted month of submission from submission time
df['submit_month'].value_counts() #month with most registration
df.groupby(["submit_month","Package"])["AgeRange"].count() #each package registration breakdown by month
#Plotting month of submission vs package using matplotlib
x=df['Package']
y=df['submit_month']
plt.barh(x,y)
#adding titles and labels
plt.title("Month of Submission vs Package")
plt.xlabel("Month")
plt.ylabel("Package")







```
